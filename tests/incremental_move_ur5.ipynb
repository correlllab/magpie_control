```python
# zero is at 0.072 on the z-axis
# 24cm from home to floor
home = np.array([[-1, 0, 0, -0.337],
                 [ 0, 1, 0, -0.497],
                 [ 0, 0,-1,  0.312],
                 [ 0, 0, 0,  1.   ]])
```

```python
robot.moveL(home)
```

```python
new = home.copy()
# make sure r*inc <= 0.23
r = 4
inc = 0.05
for i in range(r):
    new[2][3] -= inc
    robot.moveL(new)
    time.sleep(2)
    print(robot.get_tcp_pose())
```
